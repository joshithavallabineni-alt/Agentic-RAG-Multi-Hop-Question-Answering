```python
#function which helps the document making function
def extract_data(data):

  title_list = []
  sentence_list = []
  question_list =[]

  for element in data:
    question_list.append(element["question"])
    for paragraph in element["context"]:
      title_list.append(paragraph["title"])
      sentence_list.append(paragraph["sentence"])


  return title_list, sentence_list, question_list
```

```python
def make_document(title, sentence):

    docs =[]

    if (len(title)) == (len(sentence)):
        for i in range(len(sentence)):
            paragraph = "".join(sentence[i])
            docs.append(": ".join([title[i], paragraph]))

    else:
        print("Length mismatch")

    return docs
```

```python
#Search tool for retrieving the best document

def Search(query, vectorizer, vectorized_matrix, docs):
    query_vector = vectorizer.transform([query])
    cos_similarity = cosine_similarity(query_vector, vectorized_matrix)

    best_doc_index = 0

    for i in range(cos_similarity.shape[0]):
        best_doc_index = np.argmax(cos_similarity[i])


    best_doc = docs[best_doc_index]

    return best_doc

```
